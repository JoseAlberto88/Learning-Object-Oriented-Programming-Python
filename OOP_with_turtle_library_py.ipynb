## Why `turtle` is a good first step into OOP

Before learning to write your own classes, `turtle` is a nice way to see
OOP *used* rather than explained. Every turtle you create is an **object**, an instance of the `Turtle` class the library already defines for you, and it works exactly the way your own classes will later:

- `turtle.Turtle` is the **class** (the blueprint).
- `t = turtle.Turtle()` creates an **object** (an instance of that class).
- `t.forward(50)`, `t.left(90)` are **methods**: actions the object knows
  how to perform.
- Its color, position, and heading are **attributes**: state that belongs
  to that specific object.
- Create a second turtle, `t2 = turtle.Turtle()`, and it has its own
  independent position and color. Changing `t2` never affects `t`. That's
  the same idea as two objects of the same class keeping separate state
  (encapsulation), just before you've written a single `class` keyword
  yourself.

### Getting started

```python
import turtle

screen = turtle.Screen()   # the window/canvas

```

### Common methods (actions the object performs)

| Method | What it does |
|---|---|
| `forward(units)` / `fd()` | Move forward |
| `backward(units)` / `bk()` | Move backward |
| `left(degrees)` / `lt()` | Turn left |
| `right(degrees)` / `rt()` | Turn right |
| `penup()` / `pu()` | Lift the pen (move without drawing) |
| `pendown()` / `pd()` | Lower the pen (drawing resumes) |
| `goto(x, y)` | Jump straight to a coordinate |
| `setheading(angle)` | Point in an exact direction |
| `circle(radius)` | Draw a circle |
| `color(name)` / `pencolor(name)` | Set drawing color |
| `begin_fill()` / `end_fill()` | Start/stop filling a shape with color |
| `write(text)` | Write text on the canvas |
| `speed(n)` | Set drawing speed (0 = fastest, 1–10 slow to fast) |
| `hideturtle()` / `showturtle()` | Hide/show the turtle cursor |
| `reset()` / `clear()` | Clear the drawing |

### Common attributes (state the object holds)

| Read with | Represents |
|---|---|
| `t.pos()` / `t.xcor()` / `t.ycor()` | Current position |
| `t.heading()` | Current direction, in degrees |
| `t.isdown()` | Whether the pen is currently down |
| `t.isvisible()` | Whether the turtle cursor is shown |

### The OOP connection

When you draw two shapes with two separate turtle objects, say `t1` in
red and `t2` in blue, you're already practicing the core OOP habit:
each object owns its own attributes, and you interact with it only
through its methods, never by reaching in and changing its internals
directly. That's the exact mental model classes are built around.

````python
# The first exercise

clearscreen()
setup(400,400)

colors = ["red", "yellow", "blue", "green", "orange", "black"]

p = Turtle()
p.width(10)

for i in range(6):
  p.color(colors[i])
  p.forward(100)
  p.left(60)
````

````python
# The Second exercise: Draw 4 solid squares all different colors in different locations.

from random import randint

clearscreen()
setup(500,500)

s = Turtle()

for i in range(4):
  r = randint(0,255)
  g = randint(0,255)
  b = randint(0,255)

  color1 = (r, g, b)

  x = randint(-100,100)
  y = randint(-100,100)

  s.penup()
  s.goto(x,y)
  s.pendown()

  s.color(color1)
  s.begin_fill()
  for i in range(4):
    s.forward(100)
    s.left(90)
  s.end_fill()
````
