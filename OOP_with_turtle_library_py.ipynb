## Why `turtle` is a good first step into OOP

Before learning to write your own classes, `turtle` is a nice way to see
OOP *used* rather than explained. Every turtle you create is an **object**, an instance of the `Turtle` class the library already defines for you, and it works exactly the way your own classes will later:

- `turtle.Turtle` is the **class** (the blueprint).
- `t = turtle.Turtle()` creates an **object** (an instance of that class).
- `t.forward(50)`, `t.left(90)` are **methods**: actions the object knows
  how to perform.
- Its color, position, and heading are **attributes**: state that belongs
  to that specific object.
- Create a second turtle, `t2 = turtle.Turtle()`, and it has its own
  independent position and color. Changing `t2` never affects `t`. That's
  the same idea as two objects of the same class keeping separate state
  (encapsulation), just before you've written a single `class` keyword
  yourself.

### Getting started

```python
import turtle

```

### Common methods (actions the object performs)

| Method | What it does |
|---|---|
| `forward(units)` / `fd()` | Move forward |
| `backward(units)` / `bk()` | Move backward |
| `left(degrees)` / `lt()` | Turn left |
| `right(degrees)` / `rt()` | Turn right |
| `penup()` / `pu()` | Lift the pen (move without drawing) |
| `pendown()` / `pd()` | Lower the pen (drawing resumes) |
| `goto(x, y)` | Jump straight to a coordinate |
| `setheading(angle)` | Point in an exact direction |
| `circle(radius)` | Draw a circle |
| `color(name)` / `pencolor(name)` | Set drawing color |
| `begin_fill()` / `end_fill()` | Start/stop filling a shape with color |
| `write(text)` | Write text on the canvas |
| `speed(n)` | Set drawing speed (0 = fastest, 1–10 slow to fast) |
| `hideturtle()` / `showturtle()` | Hide/show the turtle cursor |
| `reset()` / `clear()` | Clear the drawing |

### Common attributes (state the object holds)

| Read with | Represents |
|---|---|
| `t.pos()` / `t.xcor()` / `t.ycor()` | Current position |
| `t.heading()` | Current direction, in degrees |
| `t.isdown()` | Whether the pen is currently down |
| `t.isvisible()` | Whether the turtle cursor is shown |

### The OOP connection

When you draw two shapes with two separate turtle objects, say `t1` in
red and `t2` in blue, you're already practicing the core OOP habit:
each object owns its own attributes, and you interact with it only
through its methods, never by reaching in and changing its internals
directly. That's the exact mental model classes are built around.

# OOP Basics: Class, Object, Attribute, and Method (with `turtle`)

Object-Oriented Programming (OOP) organizes code around **objects**, bundles of data and behavior, instead of just functions operating on loose variables. Python's built-in `turtle` library is a great way to see this in action, because every turtle you draw with is literally an object.

## Class

A **class** is a *blueprint*. It defines what data an object of that type will hold and what it will be able to do, but it isn't a usable thing by itself, it's the template.

`turtle.Turtle` is a class defined inside the `turtle` library. It specifies that any turtle will have things like a position, a heading, a pen color, and a pen state (up or down), and that it will be able to move forward, turn, and change color.

```python
import turtle

# turtle.Turtle is the class, the blueprint for "a turtle"
print(type(turtle.Turtle))  # <class 'type'>
```

## Object

An **object** (also called an **instance**) is a concrete thing built from a class. Creating one is called **instantiation**. Each object gets its own independent copy of the data described by the class.

```python
import turtle

# Each of these lines creates a separate object from the Turtle class
t1 = turtle.Turtle()
t2 = turtle.Turtle()

t1.color("blue")
t2.color("red")

t1.forward(100)
t2.forward(50)
```

`t1` and `t2` are both turtles (same class), but they are two distinct objects: `t1` is blue and moved 100 pixels, `t2` is red and moved 50. Changing `t1`'s color never affects `t2`, that separation of state is exactly what "object" means.

## Attribute

An **attribute** is a piece of data stored on an object, a variable that belongs to that specific instance. For a turtle, its current position, heading, color, and pen size are all attributes: values it carries around and that can differ from one turtle to the next.

```python
import turtle

t = turtle.Turtle()
t.color("green")
t.pensize(3)

print(t.pos())     # attribute-like state: current (x, y) position
print(t.heading())  # attribute-like state: current direction in degrees
```

Under the hood, `turtle.Turtle` stores this state internally (`_position`, `_heading`, pen color, etc.) and exposes it through methods like `.pos()` and `.heading()` rather than plain public attributes — but conceptually, this is still "data that belongs to the object." If you build your own class (see below), you'll usually set attributes directly, e.g. `self.color = "green"`.

## Method

A **method** is a function that belongs to a class and acts on a specific object, using (and often changing) that object's attributes. You call it with dot notation on an instance: `object.method(arguments)`.

```python
import turtle

t = turtle.Turtle()

t.forward(100)   # method: moves t, changing its position attribute
t.right(90)      # method: turns t, changing its heading attribute
t.color("purple")  # method: changes t's pen color attribute
t.penup()        # method: changes t's pen state
```

Every one of those calls reads or updates `t`'s own internal state — never some other turtle's. That's the core link between the four concepts: the **class** defines what **attributes** and **methods** exist; each **object** gets its own attribute values; and calling a **method** on an object operates on that object's data.

## Putting it together: drawing a square

```python
import turtle

square_turtle = turtle.Turtle()   # object, built from the Turtle class
square_turtle.color("orange")     # setting an attribute via a method
square_turtle.pensize(2)          # setting another attribute via a method

for _ in range(4):
    square_turtle.forward(100)    # method: uses/changes position
    square_turtle.right(90)       # method: uses/changes heading

turtle.done()
```

## Writing your own class: subclassing `Turtle`

Where this really clicks is when you define your own class instead of only using one from a library. You can build on top of `turtle.Turtle` by **subclassing** it — inheriting its existing attributes and methods, and adding your own.

```python
import turtle

class Square(turtle.Turtle):
    """A turtle that knows how to draw a square of a given size."""

    def __init__(self, size, color):
        super().__init__()        # sets up all the normal Turtle attributes
        self.size = size          # a new, custom attribute
        self.color(color)         # reusing the inherited method to set pen color

    def draw(self):                # a new, custom method
        for _ in range(4):
            self.forward(self.size)
            self.right(90)


# Each Square below is a separate object: same class, own size/color/position
s1 = Square(size=80, color="blue")
s2 = Square(size=120, color="red")

s2.penup()
s2.goto(150, 0)
s2.pendown()

s1.draw()
s2.draw()

turtle.done()
```

A quick map of what's what in that example:

- **Class**: `Square`: the blueprint that says "a Square turtle has a `size` and knows how to `draw()`."
- **Object**: `s1` and `s2`: two independent Square turtles, each with its own position, color, and size.
- **Attribute**: `self.size` (custom, set in `__init__`) and everything inherited from `Turtle`, like position and color.
- **Method**: `draw()` (custom) and everything inherited, like `.forward()`, `.right()`, `.penup()`.

`__init__` is the **constructor**: the method that runs automatically when you create a new object (`Square(size=80, color="blue")`), and it's where you typically set up that object's starting attributes. `self` refers to "this particular object" inside the class body, it's how a method knows which object's data to read or change.

## One-line summary

- **Class**: the blueprint (`Turtle`, `Square`).
- **Object**: a specific thing built from that blueprint (`t1`, `s1`).
- **Attribute**: data stored on an object (`size`, `color`, position, heading).
- **Method**: a function tied to a class that acts on an object's data (`forward()`, `draw()`, `color()`).

####

Here are the main things we can do with this system:

1. Enroll a student -> enroll_student(student_name): If someone wants to join a course, we can add their name to the list of students in that course. When we do this, the system will tell us that the student has successfully enrolled.

2. See course details -> get_course_details(): We can ask the system to show us information about a course. This includes the course name, the name of the instructor, and the names of all the students who have enrolled.

3. Mark a student as completed -> comlete_course(student_name): If a student finishes

````python
# Lets creating the blueprint with all the charcateristics

class OnlineCourse:
  def __init__(self, course, instructor):
    self.course = course
    self.instructor = instructor
    self.students = []

  def enroll_student(self, name):
    self.students.append(name)
    print(f"{name} has been enrolled in the {self.course} course")

  def course_details(self):
    print(f"Course: {self.course}")
    print(f"Instructor name: {self.instructor}")
    print(f"Enrolled students: {', '.join(self.students)}")

  def complete_course(self, name):
    if name in self.students:
      self.students.remove(name)
      print(f"{name} has completed the {self.course} course")
    else:
      print(f"{name} is not enrolled in the {self.course} course")

  def avg_grade(self, grades):
    total = sum(grades)
    average = total / len(grades)
    print(f"The average grade is: {average}")

````

# Class Inheritance in OOP

Class inheritance lets one class reuse and extend another class's code. The class being reused is the parent (or base/superclass). The new class is the child (or derived/subclass).

## Basic syntax (Python)

```python
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        return f"{self.name} makes a sound"

class Dog(Animal):
    def speak(self):
        return f"{self.name} barks"
```

`Dog` inherits from `Animal`. It gets `__init__` for free and overrides `speak()`.

```python
d = Dog("Rex")
print(d.speak())  # Rex barks
```

## Why use it

Say you're building a game with `Player`, `Enemy`, and `NPC` classes. All three need health, position, and a `move()` method. Instead of writing that logic three times, you put it in a `Character` base class and let the other three inherit from it. Fix a bug in `move()` once, and every subclass gets the fix.

## Calling the parent's method

Sometimes you want to extend a method, not replace it:

```python
class Cat(Animal):
    def speak(self):
        base = super().speak()
        return f"{base}, specifically a meow"
```

`super()` gives you access to the parent class, so you're not stuck rewriting logic that already works.

## Multiple levels

Inheritance can chain: `Puppy` could inherit from `Dog`, which inherits from `Animal`. Each level adds or overrides behavior without touching the layer below it.

## Common pitfall

Deep inheritance chains (four or five levels) get hard to trace. If you find yourself asking "wait, where is this method actually defined?", that's usually a sign to flatten the hierarchy or switch some of it to composition (one class holding an instance of another, instead of inheriting from it).


````python
# Let's make an example with a class Animal

class Animal:
  def __init__(self, region, animal_type, color, lethal):
    self.region = region
    self.animal_type = animal_type
    self.color = color
    self.lethal = lethal

  def animal_bio(self):
    print("Animal Passport:")
    print(f"Found in: {self.region}")
    print(f"Species: {self.animal_type}")
    print(f"Color: {self.color}")
    print(f"Dangerous: {self.lethal}")


class Clinic(Animal):
  def animal_info(self):
    print(f"This is a {self.animal_type} from {self.region}")

  def search(self, animals):
    region = input("Enter a region: ").lower()
    for animal in animals:
      if animal.region == region:
        print(f"Species: {animal.animal_type}")

animals = []
amt_animals = int(input("Enter number of animals: "))
for _ in range(amt_animals):
  region = input("Enter a region: ").lower()
  species = input("Enter a species: ").lower()
  color = input("Enter a color: ").lower()
  lethal = input("It is dangerous: ").lower()
  lethal = lethal == "yes"

  animal = Animal(region, species, color, lethal)
  animals.append(animal)

clinic = Clinic("asia", "tiger", "orange", True)

clinic.animal_bio()
clinic.animal_info()
clinic.search(animals)
````

### Title: Library Management System

Description:

This code represents a library management system that allows you to manage a collection of books. The systems provides functionalities for adding books, displaying all books, searching for books by author, and searching for books within a specific year range.

#####

Book Class:

The Book Class serves as the base class for different types of books.

- It has attributes for title, author, and year.

- The `display_info()` method is used to print the title, author, and year of a book.

###

FictionBook Class:

The FictionBook Class is a subclass of the book class. It adds an additional attribute called genre.

- The `display_info()` method is overriden to include the genre when printing book information.

###

NonFictionBook Class:

The NonFictionBook Class is another subclass of the Book class. It adds an attribute called topic.

````python
# Let's construst the code with the requerements

class Book:
  def __init__(self, title, author, year):
    self.title = title
    self.author = author
    self.year = year

  def info(self):
    print(f"Title: {self.title}")
    print(f"Author: {self.author}")
    print(f"Year: {self.year}")

class FictionBook(Book):
  def __init__(self, title, author, year, genre):
    super().__init__(title, author, year)
    self.genre = genre

  def info(self):
    super().info()
    print(f"Genre: {self.genre}")

class NonFictionBook(Book):
  def __init__(self, title, author, year, topic):
    super().__init__(title, author, year)
    self.topic = topic

  def info(self):
    super().info()
    print(f"Topic: {self.topic}")


class Library:
  def __init__(self):
    self.books = []

  def add_book(self, book):
    self.books.append(book)

  def display_books(self):
    for book in self.books:
      book.info()
      print()

  def search_author(self, author):
    found_books = []
    for book in self.books:
      if book.author == author:
        found_books.append(book)
    if found_books:
      print(f"Books by {author}:")
      for book in found_books:
        book.info()
        print()
    else:
      print(f"No books by {author} found.")

  def search_year(self, start_year, end_year):
    found_books = []
    for book in self.books:
      if start_year <= book.year <= end_year:
        found_books.append(book)
    if found_books:
      print(f"Books between {start_year} and {end_year}:")
      for book in found_books:
        book.info()
        print()
    else:
      print(f"No books between {start_year} and {end_year} found")


library = Library()

book1 = FictionBook("Harry Potter and the Philosopher's Stone", "J.K. Rowling", 1997, "Fantasy")
book2 = FictionBook("To Kill a Mockingbird", "Harper Lee", 1960, "Classic")
book3 = NonFictionBook("The Alchemist", "Paulo Coelho", 1988, "Adventure")
book4 = NonFictionBook("The Da Vinci Code", "Dan Brown", 2003, "Mystery")

library.add_book(book1)
library.add_book(book2)
library.add_book(book3)
library.add_book(book4)

library.display_books()

library.search_author("J.K. Rowling")
library.search_year(1990, 2000)
````
